```python
from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator
import matplotlib.pyplot as plt


shots=[100,1000,10000]

prob_zero=[]
prob_one=[]


for s in shots:

    qc=QuantumCircuit(1,1)

    qc.h(0)

    qc.measure(0,0)


    simulator=AerSimulator()

    result=simulator.run(
        qc,
        shots=s
    ).result()


    counts=result.get_counts()

    print("Shots:",s)
    print(counts)


    prob_zero.append(
        counts.get('0',0)/s
    )

    prob_one.append(
        counts.get('1',0)/s
    )



plt.plot(
    shots,
    prob_zero,
    marker="o",
    label="0 probability"
)


plt.plot(
    shots,
    prob_one,
    marker="o",
    label="1 probability"
)


plt.axhline(
    0.5,
    linestyle="--",
    label="Ideal 0.5"
)


plt.xlabel("Shots")
plt.ylabel("Probability")

plt.title(
"Hadamard State Probability Convergence"
)

plt.legend()
plt.grid()

plt.show()
```
