```python
import matplotlib.pyplot as plt



lengths = list(range(2,15))


classical_queries = []

simon_queries = []



for n in lengths:


    classical_queries.append(
        2**n
    )


    simon_queries.append(
        n
    )



plt.figure(figsize=(8,5))



plt.plot(
    lengths,
    classical_queries,
    marker="o",
    label="Classical Brute Force"
)



plt.plot(
    lengths,
    simon_queries,
    marker="o",
    label="Simon's Algorithm"
)



plt.xlabel(
    "Secret String Length"
)



plt.ylabel(
    "Number of Oracle Queries"
)



plt.title(
    "Simon Algorithm Speedup Comparison"
)



plt.legend()


plt.grid()


plt.show()
```
