```python
from qiskit import QuantumCircuit, transpile
from qiskit_aer import Aer


secret = "110"


def create_oracle(secret):

    n = len(secret)

    qc = QuantumCircuit(2*n, n)


    for i, bit in enumerate(secret):

        if bit == "1":

            qc.cx(i, n+i)


    return qc



n = len(secret)


qc = QuantumCircuit(2*n, n)



# Step 1: Superposition

for i in range(n):

    qc.h(i)



qc.barrier()



# Step 2: Oracle

oracle = create_oracle(secret)

qc.compose(
    oracle,
    inplace=True
)



qc.barrier()



# Step 3: Apply Hadamard again

for i in range(n):

    qc.h(i)



# Step 4: Measurement

qc.measure(
    range(n),
    range(n)
)



print("Simon's Algorithm Circuit")
print("-------------------------")

print(qc.draw())



backend = Aer.get_backend(
    "qasm_simulator"
)



job = backend.run(
    transpile(qc, backend),
    shots=1024
)



result = job.result()



print("\nMeasurement Results")

print(
    result.get_counts()
)
```
