```python
import numpy as np


secret = "1011"



def gf2_solver(matrix):

    matrix = np.array(matrix, dtype=int)

    rows, cols = matrix.shape

    pivot = 0


    for col in range(cols):

        # Stop if all rows are used
        if pivot >= rows:
            break


        # Find pivot row
        for row in range(pivot, rows):

            if matrix[row][col] == 1:

                matrix[[pivot, row]] = matrix[[row, pivot]]

                break

        else:
            continue



        # Eliminate other rows
        for row in range(rows):

            if row != pivot and matrix[row][col] == 1:

                matrix[row] = matrix[row] ^ matrix[pivot]


        pivot += 1



    return matrix




print("Secret String:")
print(secret)



# GF(2) equations obtained from Simon measurements

equations = [

    [1,0,1,1],

    [0,1,1,0],

    [1,1,0,1]

]



print("\nOriginal GF(2) Equations")

print(
    np.array(equations)
)



print("\nReduced Matrix")

print(
    gf2_solver(equations)
)
```
