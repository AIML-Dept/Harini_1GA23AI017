```python
from itertools import product


secret = "110"


def simon_oracle(x):

    x_int = int(x, 2)
    s_int = int(secret, 2)

    result = x_int ^ s_int

    return format(min(x_int, result), "03b")



print("Simon's Oracle Truth Table")
print("--------------------------")

print("Input\tOutput")


for bits in product("01", repeat=3):

    x = "".join(bits)

    print(x, "\t", simon_oracle(x))
```
