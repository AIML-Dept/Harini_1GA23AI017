```python
def generalized_simon(secret):

    n = len(secret)

    print("----------------------------")
    print("Generalized Simon Algorithm")
    print("----------------------------")


    # Check all zero case
    if secret == "0" * n:

        print("Secret string is all zeros")
        print("No oracle queries required")

        return


    print("Secret String:", secret)
    print("Length:", n)


    print("\nSimon Algorithm executed successfully")



# Give secret string here
secret = "1011"


generalized_simon(secret)
```
