```python
import numpy as np
from scipy.stats import chisquare



quantum=np.random.choice(
[0,1],
1000
)


biased=np.random.choice(
[0,1],
1000,
p=[0.8,0.2]
)



def check(data,name):

    zero=np.sum(data==0)
    one=np.sum(data==1)


    chi,p=chisquare(
        [zero,one],
        [500,500]
    )


    print("\n",name)

    print(
    "0:",
    zero,
    "1:",
    one
    )

    print(
    "p-value:",
    p
    )


    if p>0.05:
        print("Random")
    else:
        print("Biased")



check(
quantum,
"Quantum Random Generator"
)


check(
biased,
"Classical Biased Generator"
)
```
