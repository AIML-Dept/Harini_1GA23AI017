```python
from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator


sim=AerSimulator()


# |+> state

qc1=QuantumCircuit(1,1)

qc1.h(0)

qc1.h(0)

qc1.measure(0,0)



result=sim.run(
    qc1,
    shots=1000
).result()


print(
"|+> measurement:",
result.get_counts()
)



# |-> state

qc2=QuantumCircuit(1,1)

qc2.x(0)

qc2.h(0)

qc2.h(0)

qc2.measure(0,0)



result=sim.run(
    qc2,
    shots=1000
).result()


print(
"|- > measurement:",
result.get_counts()
)
```
