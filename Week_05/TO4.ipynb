```python
import numpy as np
import matplotlib.pyplot as plt

from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator


sim=AerSimulator()


actual=[]
estimated=[]


for theta in np.linspace(0,np.pi,10):

    qc=QuantumCircuit(1,1)


    qc.ry(theta,0)

    qc.measure(0,0)



    result=sim.run(
        qc,
        shots=2000
    ).result()



    counts=result.get_counts()


    p1=counts.get('1',0)/2000


    estimate=2*np.arcsin(
        np.sqrt(p1)
    )


    actual.append(theta)
    estimated.append(estimate)



plt.plot(
actual,
estimated,
marker="o"
)


plt.xlabel(
"Actual Angle"
)

plt.ylabel(
"Estimated Angle"
)

plt.title(
"Quantum State Tomography"
)


plt.grid()

plt.show()
```
